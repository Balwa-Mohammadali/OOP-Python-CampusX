```python
class Point:
    def __init__(self,x,y):
        self.x_cord=x
        self.y_cord=y

    def __str__(self):
        return '<{},{}>'.format(self.x_cord,self.y_cord)

    # distance between two points
    def euclidian_distance(self,other):
        return ((self.x_cord-other.x_cord)**2+(self.y_cord-other.y_cord)**2)**0.5
    
    def distance_from_origin(self):
        return self.euclidian_distance(Point(0,0))
        # return (self.x_cord**2+self.y_cord**2)**0.5
```

```python
p1=Point(0,0)
p2=Point(10,10)
```

```python
print(p1)
print(p2)
```

```python
p1.euclidian_distance(p2)
```

```python
p2.distance_from_origin()
```

```python
class Line:
    def __init__(self,A,B,C):
        self.A=A
        self.B=B
        self.C=C

    def __str__(self):
        return '{}x + {}y + {}'.format(self.A,self.B,self.C)

    # put values of A,B,C and x1,y1 in equation and if:Ax1+By1+C==0 then:"point lies on line"
    def point_on_line(line,point):
        if line.A*point.x_cord+line.B*point.y_cord+line.C==0:
            return "point lies on line"
        else:
            return "point doesn't lies on line"

    # find sortest distance from point to line : |Ax1+By1+C|/√(A^2+B^2)
    def sortest_distance(line,point):
        return abs(line.A*point.x_cord + line.B*point.y_cord +line.C)/(line.A**2+line.B**2)**0.5
```

```python
l1=Line(1,1,-2)
p3=Point(1,2)
print(l1)
print(p3)
```

```python
print(l1.point_on_line(p3))
print(l1.sortest_distance(p3))
```

```python
l4=Line(1,1,-2)
p4=Point(1,1)
print(l4)
print(p4)
print(l4.point_on_line(p4))
print(l4.sortest_distance(p4))
```
