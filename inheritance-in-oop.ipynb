```python
# Parent class
class User:
    def __init__(self):
        self.name='harry'

    def login(self):
        return 'login successful'

# Child class
class Student(User):

    def enroll(self):
        return 'enroll successful'
```

##  Method Overriding 

**when both parent and child have same method then child's method is run that is called method overriding**

```python
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("parent: Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("child: Buying a smartphone")

s=SmartPhone(20000, "Apple", 13)

s.buy()
```

# Super() --> super keyword

**1. super cannot access variables**

**2. super cannot be used outside the class**

**3. super() is used in child class**

```python
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka buy method
        super().buy()

s=SmartPhone(20000, "Apple", 13)

s.buy()
```

```python
# super -> constuctor
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print('Inside smartphone constructor')
        super().__init__(price, brand, camera)
        self.os = os
        self.ram = ram
        print ("Inside smartphone constructor")

s=SmartPhone(20000, "Samsung", 12, "Android", 2)

print(s.os)
print(s.brand)
```
