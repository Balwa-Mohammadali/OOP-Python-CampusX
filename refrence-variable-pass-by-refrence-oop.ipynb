```python
class Person:
    def __init__(self,name,gender):
        self.name=name
        self.gender=gender
```

```python
# outside the class -> function
def greet(person):
    print(id(person))
    print("i'm", person.name, ",my gender is" ,person.gender)
    p2=Person('mery','female')
    return p2

p1=Person('harry','male')
print(id(p1))
x=greet(p1)   
print(x.name) 
print(x.gender)
```
