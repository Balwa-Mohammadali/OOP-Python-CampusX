# 1.Encapsulation: private variable , getter and setter method

```python
class ATM:

    def __init__(self):
        self.__balance = 0       # private attribute

    # Setter
    def set_balance(self, amount):
        if amount >= 0 and type(amount)==int:
            self.__balance = amount
        else:
            print('abe sale')
    # Getter
    def get_balance(self):
        return self.__balance


```

```python
atm1=ATM()
```

```python
atm1.get_balance()
```

```python
atm1.set_balance(1000)
```

```python
atm1.get_balance()
```

# 2.Collection of object

```python
class person:
    def __init__(self,name,gender):
        self.name=name
        self.gender=gender

p1=person('merry','female')
p2=person('yoyo','male')
p3=person('kk','male')

L=[p1,p2,p3]

for i in L:
    print(i.name,i.gender)
    
```

```python
d={'p1':p1,'p2':p2,'p3':p3}

for i in d:
    print(d[i].name,d[i].gender)
```

# 3. Static variable (class variable)

**(i) instance variable ka value har object ke liye alag alag hota hai and static variable ka value har object ke liye same hota hai**

**(ii) instance variable object ka hota hai or static variable class ka hota hai**

```python
class Atm:
    # static variable
    __counter=1

    def __init__(self):
        self.cid=Atm.__counter
        Atm.__counter=Atm.__counter+1

    # static method -------> utility method
    @staticmethod
    def get_counter():
        return Atm.__counter

    @staticmethod
    def set_counter(amount):
        Atm.__counter=amount
        
```

```python
c1=Atm()
```

```python
c2=Atm()
```

```python
c3=Atm()
```

```python
Atm.get_counter()
```
