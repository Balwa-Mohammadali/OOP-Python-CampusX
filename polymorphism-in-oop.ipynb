## 2. Method overloading :

**mutiple method with same name in a single class, but thier output is differnt depend on thier input**

```python
# 2. method overloading : 
#      so this is method overloading but python doesn't support it
class shape:
    def area(self,radii):
        return 3.14*radii*radii

    def area(self,L,B):
        return L*B
        
s=shape()
s.area(2)
s.area(3,4)
```

```python
# so we can implement method overloading in python using default parameter
class Shape:

  def area(self,a,b=0):
    if b == 0:
      return 3.14*a*a
    else:
      return a*b

s = Shape()

print(s.area(2))
print(s.area(3,4))
```
